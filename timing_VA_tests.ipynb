```python
for case_idx, case_dict in enumerate(conf_cases):
    case_number = case_idx+1

    # Conf params
    f_start_list = case_dict['f_start_list']
    f_stop_list = case_dict['f_stop_list']
    points_list = case_dict['points_list']
    ifbw = case_dict['ifbw']
    n_sweeps = case_dict['n_sweeps']
    average = case_dict['average']
    n_sparams = case_dict['n_sparams']

    if average:
        n_average = n_sweeps
    else:
        n_average = 1

    if n_sparams == 3:
        sparam_list = ['S21', 'S31', 'S41']

    elif n_sparams == 6:
        sparam_list = ['S21', 'S31', 'S41', 'S22', 'S33', 'S44']

    elif n_sparams == 12:
        sparam_list = ['S21', 'S31', 'S41', 'S22', 'S33', 'S44', 'S32', 'S42', 'S23', 'S43', 'S24', 'S34']

    else:
        print('Error in s_params value')
    
    vna.configure_multiple_segmented_sparams(sparam_list=sparam_list, f_start_list=f_start_list, f_stop_list=f_stop_list, points_list=points_list, ifbw=ifbw, pow_dbm_list=pow_dbm_list)
    vna_freqs = vna.get_freq_vector()

    conf_params = {
        'ifbw': ifbw,
        'pow_dbm_list': pow_dbm_list,
        'n_sweeps': n_sweeps,
        'n_average': n_average,
        'sparam_list': sparam_list,
        'f_start_list': f_start_list,
        'f_stop_list': f_stop_list,
        'points_list': points_list
    }

    trace_list_all = [v+1 for v in range(n_sparams)] # List of all traces to measure
    trace_list_bistatic = [1, 2, 3] # List of bistatic traces to measure

    # Measure
    vna.set_sweeps(sweeps=n_sweeps)
    if average:
        vna.set_average(avg_sweeps=n_average)
        row_sparams = np.zeros((va_Nh, 1, len(trace_list_all), freq_points), dtype=np.complex64)
    else:
        vna.unset_average()
        row_sparams = np.zeros((va_Nh, n_sweeps, len(trace_list_all), freq_points), dtype=np.complex64)
    
    freq_points =  len(vna_freqs)
    timestamp_list = []

    # The 'with' statement automatically handles connection and safe disconnection
    with TriaxialPositioner(connection_str=CONN_STR) as pos:
        pos.set_velocity(min_speed=200, max_speed=700, slope=2000)
        pos.home_all_axis()
        pos.define_global_coord(x=0.0, y=0.0, z=0.0, unit='mm')

        # Executing a 2D Grid Scan using Generators
        grid_iterator = pos.generate_grid_2d(
            axis1='Y',
            axis2='Z',
            points1=va_Nh, 
            points2=va_Nv, 
            spacing1=va_deltah, 
            spacing2=va_deltav, 
            unit='mm'
        )
        
        print(f"{time_now()} Started Case {case_number}")
        for step in grid_iterator:
            step_i = step['i']
            step_j = step['j'] + va_vert_pos * va_Nv
 
            # Measure and collect sparams
            if step_i == 0:
                # All measurements both mono and bistatic (S21, S31, S41, S22, S33, S44)
                # 1. Measure with average
                sparams = vna.measure_multiple_traces_sparams(trace_list = trace_list_all)

                # Save the sparams in the row file
                row_sparams[step_i, :, :] = sparams

                timestamp_list.append(get_timestamp())
            else:
                # Only bistatic measurements (S21, S31, S41)
                sparams = vna.measure_multiple_traces_sparams(trace_list = trace_list_bistatic)

                # Save the sparams in the row file
                row_sparams[step_i, 0:3, :] = sparams

                timestamp_list.append(get_timestamp())

            # Save file for each row
            if step_i == va_Nh - 1: # last horizontal position of the row
                # Current Row Filename
                filename_row = f"{folder_name}/setup_{case_number}_nv_{step_j}"
                np.savez(filename_row, 
                    row_sparams=row_sparams, 
                    conf_params=conf_params, 
                    timestamp_list=timestamp_list, 
                    bs_params=bs_params, 
                    ue0_params=ue0_params, 
                    ue1_params=ue1_params, 
                    ue2_params=ue2_params)
                time.sleep(1)

        print(f"{time_now()} Finished Case {case_number}")
        # Return Home
        pos.move_axes_absolute(x=0.0, y=0.0, z=0.0, wait_to_finish=True, unit='mm')


```
